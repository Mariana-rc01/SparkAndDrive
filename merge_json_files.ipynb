```python
import json
import glob
import os

def merge_jsons(folder_path, output_path):
    files = glob.glob(os.path.join(folder_path, "*.json"))
    merged = {"runs": []}

    for file in files:
        with open(file, "r") as f:
            data = json.load(f)
            merged["runs"].extend(data.get("runs", []))

    # Ordena por timestamp
    merged["runs"].sort(key=lambda x: x["timestamp"])

    with open(output_path, "w") as f:
        json.dump(merged, f, indent=2)

    print(f"Merged {len(files)} files → {output_path} ({len(merged['runs'])} runs)")

merge_jsons("data/results/models",  "data/results/models_merged.json")
merge_jsons("data/results/queries", "data/results/queries_merged.json")
```
